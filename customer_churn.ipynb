```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import warnings
warnings.filterwarnings("ignore")

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from imblearn.over_sampling import SMOTENC
```

```python
df = pd.read_csv("telco_churn.csv")

df = df.drop(columns=['Unnamed: 0', 'customerID'], errors='ignore')
```

```python
df.head()
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
df['Churn'].value_counts()
```

```python
counts = df['Churn'].value_counts()

print(f"Class Imbalance Ratio: {counts['No'] / counts['Yes']:.2f}:1")

#class imbalance ratio is 2.67:1 for every 1 customer who churned, approximately 2.67 customers stayed.
```

```python
# creating two sets of visualizations:
# 1. Numerical feature distributions.
# 2. Customer churn by contract type.
fig, axes = plt.subplots(1, 3, figsize=(16,4))

for idx, col in enumerate(["tenure", "MonthlyCharges","TotalCharges"]):
    sns.histplot(df[col], kde=True, ax=axes[idx], color='skyblue')
    axes[idx].set_title(f'{col} Distribution', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

#churn by contract

fig, ax = plt.subplots(1,1, figsize=(10, 5))
contract_churn = df.groupby(['Contract', 'Churn']).size().unstack(fill_value=0)
contract_churn.plot(kind='bar', stacked=False, ax=ax, color=['#2ecc71', '#e74c3c'])
ax.set_title('Churn by Contract type', fontsize=14, fontweight='bold')
ax.set_xlabel('Contract Type')
ax.set_ylabel('Count')
ax.legend(['No Churn', 'Churn'])
ax.set_xticklabels(ax.get_xticklabels(), rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Encode binary columns

binary_cols = ['SeniorCitizen', 'Partner', 'Dependents', 'PhoneService', 'PaperlessBilling', 'Churn']

for col in binary_cols:
    df[col] = df[col].astype('string').str.strip().str.lower().map({'yes': 1, 'true': 1, 'no': 0, 'false': 0})

df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})

categorical_cols = ['MultipleLines','InternetService', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection','TechSupport','StreamingTV','StreamingMovies','Contract','PaymentMethod']

encoders = {}

for col in categorical_cols:
    encoder = LabelEncoder()
    df[col] = encoder.fit_transform(df[col].astype('string').fillna('__MISSING__'))
    encoders[col] = encoder
```

```python
df = df.dropna(subset=['Churn'])

X = df.drop('Churn', axis=1)
Y = df['Churn']

x_train, x_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    stratify=Y,
    random_state=42
)

x_train.shape[0] 
```

```python
x_test.shape[0]
```

```python
numeric_feature_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
categorical_feature_cols = [col for col in x_train.columns if col not in numeric_feature_cols]

x_train = x_train.copy()
x_test = x_test.copy()

for col in numeric_feature_cols:
    x_train[col] = pd.to_numeric(x_train[col], errors='coerce')
    x_test[col] = pd.to_numeric(x_test[col], errors='coerce')
    median = x_train[col].median()
    x_train[col] = x_train[col].fillna(median)
    x_test[col] = x_test[col].fillna(median)

for col in categorical_feature_cols:
    most_frequent = x_train[col].mode().iloc[0]
    x_train[col] = x_train[col].fillna(most_frequent)
    x_test[col] = x_test[col].fillna(most_frequent)

categorical_feature_indices = [x_train.columns.get_loc(col) for col in categorical_feature_cols]
smote = SMOTENC(categorical_features=categorical_feature_indices, random_state=42)

x_train_balanced, y_train_balanced = smote.fit_resample(x_train, y_train)

# Before SMOTE
dict(y_train.value_counts())
```

```python
#After Smote
dict(pd.Series(y_train_balanced).value_counts())
```

```python
xgb_model = XGBClassifier(n_estimators=100, max_depth=5, learning_rate=0.1, random_state=42)

xgb_model.fit(x_train_balanced, y_train_balanced)
```

```python
y_train_pred = xgb_model.predict(x_train)
y_test_pred = xgb_model.predict(x_test)

train_acc = accuracy_score(y_train, y_train_pred)
test_acc = accuracy_score(y_test, y_test_pred)
```

```python
print('Training Accuracy', train_acc)
print('Test Accuracy', test_acc)
```

```python
cm = confusion_matrix(y_test, y_test_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['No Churn', 'Churn'], yticklabels=['No Churn', 'Churn'])

plt.title('Confusion Matrix - XGBoost', fontsize=14, fontweight='bold')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()
```

```python
feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': xgb_model.feature_importances_
}).sort_values('Importance', ascending=False)

feature_importance.head(10)

plt.figure(figsize=(10,8))

plt.barh(feature_importance['Feature'][:15], feature_importance['Importance'][:15], color='coral')
plt.xlabel('Importance')
plt.title('Top 15 Feature Importances - XGBoost', fontsize=14, fontweight='bold')
plt.gca().invert_yaxis()
plt.show()
```
